```python
import os
from PIL import Image
import pillow_heif
```

```python
# HEIC 열 수 있도록 등록
pillow_heif.register_heif_opener()

root = r"C:\temp_test\260224-ViT\data\train"

def is_ftyp(path):
    try:
        with open(path, "rb") as f:
            head = f.read(32)
        return b"ftyp" in head
    except:
        return False
```

```python
converted = 0
deleted = 0
failed = []

for dirpath, _, filenames in os.walk(root):
    for fn in filenames:
        if fn.lower().endswith((".jpg", ".jpeg")):
            path = os.path.join(dirpath, fn)

            if is_ftyp(path):  # 겉은 jpg인데 실제는 HEIC/HEIF 계열
                try:
                    img = Image.open(path).convert("RGB")

                    base, _ = os.path.splitext(path)
                    new_path = base + "_converted.jpg"

                    img.save(new_path, format="JPEG", quality=95)
                    
                    os.remove(path) # 저장 성공하면 원본 삭제

                    converted += 1
                    deleted += 1
                    print(f"[OK] Converted & Deleted: {path}")

                except Exception as e:
                    failed.append((path, str(e)))
                    print(f"[FAIL] {path} | {e}")
print(converted, deleted)
if failed:
    print("\n실패 파일 목록:")
    for p, e in failed:
        print(p, "|", e)
```
